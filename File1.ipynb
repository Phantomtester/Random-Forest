```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import seaborn as sns
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("winequality-red.csv")
df.head()
```

```python
sns.catplot(data=df, x="quality", hue="quality", kind="count")
```

```python
plt.figure(figsize=(5,5))
plt.bar(df["quality"], df["pH"])
plt.xlabel("Wine Quality")
plt.ylabel("pH")
```

```python
# correlation   -- 1. Positive   2. Negative
crr = df.corr()
plt.figure(figsize=(10,10))
sns.heatmap(crr, annot=True, cmap="Blues", square=True, fmt=".1f", annot_kws={"size": 10})
```

```python
X = df.drop("quality", axis=1)
X
```

```python
y = df["quality"].apply(lambda y: 1 if y >= 7 else 0)
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=3)
rc = RandomForestClassifier(criterion="entropy")
rc.fit(X_train, y_train)
y_pred = rc.predict(X_test)
acc = accuracy_score(y_test, y_pred)
acc
```

Predictive system

```python
input_data = (7.3,0.65,0.0,1.2,0.065,15.0,21.0,0.9946,3.39,0.47,10.0)

#transforming input into numpy arr
arr = np.asarray(input_data)

#reshaping the data for pridiction
rs_data = arr.reshape(1,-1)

#prediction
prediction = rc.predict(rs_data)
print(prediction)
```
